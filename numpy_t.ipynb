```python
import numpy as np

print(np.__version__)
```

```python
ar=np.array([1,2,3,4,5])
print(ar,type(ar),ar.shape)

ar1=np.array([1,2,3,4,5,'prime'])
print(ar1,type(ar1),ar1.shape)
```

```python
ar2=np.zeros((3,3))
print(ar2,type(ar2),ar2.shape)
```

```python
ar3=np.ones((2,3))
print(ar3,type(ar3),ar3.shape)
```

```python
ar4=np.full((2,3),5)
print(ar4,type(ar4),ar4.shape)
```

```python
ar5=np.eye(3)
print(ar5,type(ar5),ar5.shape)
```

```python
ar6=np.arange(1,11)
ar7=np.arange(1,11,2)
print(ar6,type(ar6),ar6.shape)
print(ar7,type(ar7),ar7.shape)
```

```python
ar8=np.linspace(1,100,4)
print(ar8,type(ar8),ar8.shape)
```
