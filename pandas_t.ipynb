```python
import pandas as pd
df=pd.read_csv('messy_sample_data.csv')
df
```

```python
df.isnull().sum()
```

```python
df.dropna()
df.dropna(axis=1) #drop columns with missing values
```

```python
df.fillna(0) #fill missing values with 0
```

```python
df["age"].fillna(df["age"].mean()) #fill missing values in Age column with mean age
```

```python
cleaned_data=df.copy()
cleaned_data["age"].fillna(cleaned_data["age"].mean())
```

```python
cleaned_data=df.ffill() #forward fill - fills missing values with the last valid observation
cleaned_data=df.bfill() #backward fill - fills missing values with the next valid observation
```

```python
#Handles Duplicates Valuse
df.duplicated() #returns a boolean Series indicating whether each row is a duplicate of a previous row
```

```python
df["temperature"].fillna(df["temperature"].mean()) #fill missing values in temperature column with mean value of that column
```
