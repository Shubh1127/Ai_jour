```python
import json
```

```python
#Load the data from the JSON file
def load_data(filename):
    with open(filename, 'r') as f:
        data = json.load(f)
    return data
```

```python
data=load_data("log.json")
def clean_data(data):
    # clean_data(data)
    cleaned_user=[]
    unique_users=set()
    text_to_num={'one':1,'two':2,'three':3,'four':4,'five':5}
    for user in data:
        raw_rating=str(user["rating"]).strip().lower()
        # print(raw_rating)
        if(raw_rating in text_to_num):
                user["rating"]=text_to_num[raw_rating]
        else:
             user["rating"]=raw_rating
        # print(raw_rating)

        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None
        
        if(user["name"] in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_user.append(user)
        

    return cleaned_user
print(clean_data(data))
```

```python
def get_insights(data):
    total_ratings=0
    for user in data:
        total_ratings+=float(user["rating"])
    average_rating=total_ratings/len(data)
    return average_rating
```

```python
data=load_data("log.json")
cleaned_data=clean_data(data)
print(get_insights(cleaned_data))
```
