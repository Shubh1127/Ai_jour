```python
#Probability
#Probability is a measure of the likelihood that a particular event will occur (Ranges from 0 to 1).

#1. Theoretical Probability

#2. Empirical (Experimental) Probability
```

```python
#Important Terms

# Sample Space (S) - all possible outcomes

# Event (E) - a subset of outcomes

#  P(E) - P(E) is the probability of event E


# P(E) = Number of favorable outcomes / Total number of outcomes = n(E) / n(S)
```

```python
#Types of Events

 #Certain Event: An event that always happens.

 #Impossible Event: An event that cannot happen.

 #Probability of the entire sample space S is 1: P(S)=1

#Probability of an impossible event (no outcomes) is 0: P(Ø)=0
```

```python
#Intersection & Union Events

#P(AnB) - Probability that A and B both occur. [intersection]

#P(AUB) - Probability that A or B both occur (i.e. atleast one) [union]
```

```python

#Mutually Exclusive (Disjoint) Events: Two events that cannot happen at the same time.

#Independent Events: The outcome of one event does not affect the probability of the other.

#Dependent Events: The outcome of one event affects the probability of the other.

#Complementary Events: Complement event is if one occurs exactly when the other does not.
```

```python
#Rule1 - Complementary Rule

#If A is an event, then its complement Ac (event not happening) has probability:

#P(Ac) = 1-P(A)
```

```python
#Rule2 - Addition Rule

#Mutually exclusive events
#       If A and B cannot happen together: P(A or B) = P(A) + P(B)

#Non-mutually exclusive events
#       If A and B can happen together: P(A or B) = P(A) + P(B) - P(AnB)
```

```python
#Rule3 - Multiplication Rule

#Independent events
#If A and B are independent (one doesn't affect the other): P(A and B) = P(A) x P(B)

#Dependent events
#If A and B are dependent (affect each other): P(A and B) = P(A) x P(B | A)
```

```python
#Conditional Probability

#It is the probability of an event B occurring given that another event A has already happened.

#We write it as:

#   P(B | A) = P(A ∩ B) / P(A)
```

```python

#Law of Total Probability
# #It is used to find the probability of an event B by breaking it into several mutually exclusive cases.

#P(B) = P(B | A1)* P(A1) + P(B |A2)* P(A2) + ... + P(B | An) * P(An)

#Where:

#A1,A2, ... ,An are mutually exclusive and exhaustive events (one of them must happen).

#P(B | Ai) is the probability of B given Ai occurred.
```

```python

#Bayes' Theorem

#P(A | B) = ( P(B | A) * P(A)) / P(B)

#Where:


#P(A | B) = probability of event A given event B occurred


#P(B | A) = probability of event B given event A occurred


#P(A) = probability of event A before knowing B


#P(B) = total probability of event B
```

```python
#Random Variables

#A random variable (RV) is a numerical quantity that depends on the outcome of a random experiment.

#Discrete RV - countable values (integer)-- eg coin toss , dice

#Continuous RV - any value in a range (fractions)-- eg weight, height , time  ,  
```

```python
#Mean , Meadian and Mode


# mean - average value --- sum of all values / number of values

# median - middle value --- arrange values in ascending order and find the middle value

# mode - most frequently occurring value --- find the value that appears most often in the dataset

# weighted mean - average value where each value has a different weight or importance. It is calculated by multiplying each value by its weight, summing the results, and dividing by the total weight.
```

```python
#Variance and Standard Deviation

#variance - measure of how far a set of numbers are spread out from their average value. It is calculated by taking the average of the squared differences from the mean.

#standard deviation - measure of the amount of variation or dispersion of a set of values. It is the square root of the variance.

#mean absolute deviation - average of the absolute differences between each value and the mean. It provides a measure of dispersion that is less sensitive to outliers than variance and standard deviation.
```
